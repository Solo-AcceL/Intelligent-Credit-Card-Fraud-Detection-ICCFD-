## Key design choices
- Keep the raw source immutable.
- Remove exact duplicates for the primary benchmark but document the ambiguity.
- Split chronologically (60/20/20) to avoid future-to-past leakage.
- Prioritize average precision, recall, precision and false-positive review volume over accuracy.
- Use DuckDB for SQL ingestion/profile/aggregation; use Python ML libraries for model development.
